```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

class linefollow:

  def __init__(self,input):
    self.input = input
    self.cap = cv2.VideoCapture(self.input)


  def scale(self, output_filename="output_video.mp4"):
    width = int(self.cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(self.cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = self.cap.get(cv2.CAP_PROP_FPS)


    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    # VideoWriter (const String &filename, int fourcc, double fps, Size frameSize, bool isColor=true)
    out = cv2.VideoWriter('output_video.mp4', fourcc, fps, (width, height), isColor=True)

    print(height)

    while self.cap.isOpened():
      ret, frame = self.cap.read()
      center_x = width // 2
      y = 200

      if not ret:
        break

      # convert picture to binary to analyze the line
      gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

      threshold = 100
      _, binary = cv2.threshold(gray, threshold, 255, cv2.THRESH_BINARY)
      # i dont know what this _, means

      # im confused what to do here onwards
      # evaluate each row from half to bottom

      row = binary[y, :]
      # find indices where black (line) is located
      black_locations = []
      # find the mean of the black in that given row

      for idx, bin in enumerate(row):
        if bin == 0:
          black_locations.append(idx)

      if len(black_locations) > 0:
        center_x = int(np.mean(black_locations))

      cv2.circle(frame, (center_x, y), 20, (0, 0, 255), -1) # third input is the size of the ball, don't understand why the thickness is set as -1

      out.write(frame)

      # only struggle what trying to implement cv2_imshow and turns out it dosent work well with google colab and for some reason
      # i can only show gray or binary but not out

    self.cap.release()
    out.release()
```
